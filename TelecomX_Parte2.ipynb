```python
import pandas as pd

# Carga del dataset limpio
df = pd.read_csv('/content/datos_tratados.csv')

# Vista general del dataset
print("📊 INFORMACIÓN GENERAL DEL DATASET")
print("-------------------------------------------------------------------------------------------------------------------")
print(f"Tamaño del DataSet: {df.shape[0]:,} filas x {df.shape[1]} columnas")
print("\n-------------------------------------------------------------------------------------------------------------------")

df.info()
df.describe()
# Primeras filas
df.sample(10)
```

```python
# 2. Análisis exploratorio rápido
print("------------------------ Información General ------------------------")
print(df.info())

print("\n------------------------ Estadísticas Descriptivas ------------------------")
display(df.describe())

print("\n------------------------ Verificación de valores nulos ------------------------")
print(df.isnull().sum())
```

```python
# 2. Eliminar columnas que no aportan valor (identificadores únicos)
# Nota: Ajusta 'customerID' si el nombre de la columna varía en tu CSV
columnas_a_eliminar = ['CustomerID']

df_clean = df.drop(columns=[col for col in columnas_a_eliminar if col in df.columns])

print(f"Columnas restantes: {df_clean.columns.tolist()}")
```

```python
# 3. Transforma las variables categóricas a formato numérico usando One-Hot Encoding
# Se utiliza drop_first=True para evitar la colinealidad (trampa de la variable ficticia)
df_final = pd.get_dummies(df_clean, drop_first=True)

print("Variables transformadas. Nuevas dimensiones:", df_final.shape)
df_final.head()
```

```python
# 4. Calcular la proporción de clientes que cancelaron (Churn)
# Asumiendo que la columna objetivo tras el encoding se llama 'Churn' o similar
target = 'Churn' if 'Churn' in df_final.columns else [col for col in df_final.columns if 'Churn' in col][0]

proporcion = df_final[target].value_counts(normalize=True) * 100

print(f"Análisis de la variable objetivo: {target}")
print(df_final[target].value_counts())
print(f"\nProporción:\n{proporcion}")

if proporcion.min() < 30:
    print("\n⚠️ Existe un desbalance significativo en las clases.")
else:
    print("\n✅ Las clases están relativamente balanceadas.")
```

```python
# 5. Aplica técnicas de balanceo (Oversampling con SMOTE)
from imblearn.over_sampling import SMOTE

# Separar características (X) y etiqueta (y)
X = df_final.drop(columns=[target])
y = df_final[target]

# Aplicar SMOTE
smote = SMOTE(random_state=42)
X_res, y_res = smote.fit_resample(X, y)

print("Distribución antes de SMOTE:", y.value_counts().to_dict())
print("Distribución después de SMOTE:", y_res.value_counts().to_dict())
```

```python
# 6. Escalar los datos para modelos sensibles a la distancia (KNN, SVM, Regresión Logística)
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Ajustar y transformar las características balanceadas
X_scaled = scaler.fit_transform(X_res)

# Convertir de vuelta a DataFrame para facilitar la lectura
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

print("Datos estandarizados (Media=0, Desviación Estándar=1).")
X_scaled_df.head()
```

División del conjunto de datos

Usaremos una división de 80% para entrenamiento y 20% para prueba, asegurando que el modelo tenga suficientes ejemplos para aprender.

```python
from sklearn.model_selection import train_test_split

# Definimos X (características) y y (objetivo)
# Usamos las variables balanceadas con SMOTE de la celda 6
X = X_res
y = y_res

# 1. División 80/20
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

print(f"Muestras de entrenamiento: {X_train.shape[0]}")
print(f"Muestras de prueba: {X_test.shape[0]}")
```

Modelo 1 - Regresión Logística (Requiere Normalización)


Justificación: La Regresión Logística calcula coeficientes para cada variable. Si una variable tiene un rango de 0 a 1000 (Gasto Total) y otra de 0 a 1 (Género), el modelo dará una importancia desproporcionada a la de mayor magnitud. La normalización asegura que todas las variables contribuyan equitativamente al proceso de optimización.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# 2. Creamos un Pipeline que normaliza y luego aplica el modelo
model_log = Pipeline([
    ('scaler', StandardScaler()),
    ('logistic', LogisticRegression(random_state=42))
])

# Entrenamiento
model_log.fit(X_train, y_train)

# Predicción
y_pred_log = model_log.predict(X_test)

print("--- Evaluación: Regresión Logística ---")
print(classification_report(y_test, y_pred_log))
```

Modelo 2 - Random Forest (No requiere Normalización)


Justificación: Los modelos de árboles de decisión realizan particiones basadas en umbrales (ej. "¿es el gasto > 500?"). No calculan distancias ni coeficientes, por lo que funcionan igual de bien con datos originales o escalados.

```python
from sklearn.ensemble import RandomForestClassifier

# 2. Crear el modelo de Bosque Aleatorio
model_rf = RandomForestClassifier(n_estimators=100, random_state=42, max_depth=10)

# Entrenamiento (usamos los datos sin escalar X_train directamente)
model_rf.fit(X_train, y_train)

# Predicción
y_pred_rf = model_rf.predict(X_test)

print("--- Evaluación: Random Forest ---")
print(classification_report(y_test, y_pred_rf))
```

Comparación Visual de Matrices de Confusión

La matriz de confusión es vital en Churn: nos dice cuántos clientes que se iban a ir no logramos detectar (Falsos Negativos).

Random Forest con K-Fold Cross-Validation

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score, KFold
import numpy as np

# 1. Configurar el validador cruzado (K-Fold)
# Usamos shuffle=True para barajar los datos antes de dividir
kf = KFold(n_splits=5, shuffle=True, random_state=42)

# 2. Instanciar el modelo Random Forest
# Ajustamos algunos hiperparámetros básicos para evitar overfitting
rf_kfold = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42)

# 3. Ejecutar Cross-Validation
# Evaluamos usando la métrica 'accuracy' (exactitud), pero podrías usar 'f1' o 'recall'
resultados_acc = cross_val_score(rf_kfold, X_res, y_res, cv=kf, scoring='accuracy')
resultados_f1 = cross_val_score(rf_kfold, X_res, y_res, cv=kf, scoring='f1')

# 4. Mostrar resultados
print(f"Resultados de Accuracy por Fold: {resultados_acc}")
print(f"➡️ Accuracy Promedio: {np.mean(resultados_acc):.4f} (+/- {np.std(resultados_acc):.4f})")
print(f"➡️ F1-Score Promedio: {np.mean(resultados_f1):.4f}")

# 5. Entrenar el modelo final con todos los datos de entrenamiento para predicciones futuras
rf_kfold.fit(X_train, y_train)
```

Análisis Crítico y Comparativo
1. ¿Cuál modelo tuvo el mejor desempeño?
Generalmente, Random Forest suele obtener un F1-score más alto en este tipo de datasets debido a su capacidad para capturar relaciones no lineales entre las variables (por ejemplo, cuando el riesgo de Churn sube solo en rangos específicos de precio). Sin embargo, la Regresión Logística es más interpretable para los stakeholders.

2. Overfitting y Underfitting
Overfitting (Sobreajuste): Si notas que el Accuracy en el entrenamiento es del 99% pero en la prueba cae al 75%, el modelo (especialmente el Random Forest) está memorizando los datos.

Ajuste: Podrías reducir max_depth en el Random Forest o aumentar la regularización C en la Regresión Logística.

Underfitting (Subajuste): Si ambos porcentajes son bajos (ej. 60%), el modelo es demasiado simple.

Ajuste: Podrías agregar más variables (feature engineering) o permitir que los árboles sean más profundos.

3. Importancia de las Métricas
En el problema de Churn, el Recall es a menudo más importante que la Exactitud. Es preferible tener un "Falso Positivo" (darle un descuento a alguien que no se iba a ir) que un "Falso Negativo" (perder a un cliente porque el modelo no predijo que se iría).


Análisis de Importancia de las Variables

Utilizaremos los coeficientes de la Regresión Logística y la importancia de características de Random Forest para contrastar resultados.

```python
import numpy as np

# 1. Importancia en Regresión Logística (Coeficientes)
# Los coeficientes indican la dirección y fuerza de la relación
importancia_log = model_log.named_steps['logistic'].coef_[0]
features = X.columns

# 2. Importancia en Random Forest (Gini Importance)
importancia_rf = model_rf.feature_importances_

# Crear un DataFrame comparativo
df_importancia = pd.DataFrame({
    'Variable': features,
    'Coef_Logistica': importancia_log,
    'Imp_RandomForest': importancia_rf
}).sort_values(by='Imp_RandomForest', ascending=False)

# Visualización
plt.figure(figsize=(12, 8))
sns.barplot(x='Imp_RandomForest', y='Variable', data=df_importancia.head(10), palette='viridis')
plt.title('Top 10 Variables más influyentes (Random Forest)')
plt.show()

print("Variables con mayor impacto positivo en Churn (Regresión Logística):")
print(df_importancia[['Variable', 'Coef_Logistica']].sort_values(by='Coef_Logistica', ascending=False).head(5))
```
